```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.maximize_window()

driver.get("https://edu-dewi-aulia-nurjanah-si02.odoo.com/odoo")  # URL Odoo kamu
time.sleep(3)
```

```python
# login ke Odoo
def login(username, password):

    driver.find_element(By.ID, "login").send_keys(username)
    driver.find_element(By.ID, "password").send_keys(password)
    driver.find_element(
        By.XPATH,
        "//button[contains(text(),'Log in')]"
    ).click()

    time.sleep(5)

login("065002400010@std.trisakti.ac.id", "DeyliaTrisakti8")
```

```python
# navigasi ke modul Sales
def go_to_sales_module():

    sales_module = driver.find_element(
        By.XPATH,
        "//a[contains(., 'Sales')]"
    )

    sales_module.click()

    time.sleep(5)

go_to_sales_module()
```
